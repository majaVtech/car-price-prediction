# Task - Used car price prediction
In this part of the processing, we will investigate the composition of the dataset, whether there are missing values, invalid values, extreme values, etc. We will also familiarize ourselves with all types of data that our dataset contains.

### Step 1: Loading the dataset

Here I will gather all the necessary information about the dataset to get started.

I have shown the shape of the dataset, rows and columns.
Information about the dataset, what to expect from the dataset, what data type, column name, number of columns, index range and a preview of the dataset through a few random rows.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/majaVtech/car-price-prediction/main/Data/cars.csv"
df = pd.read_csv(url)

print("Shape of the dataset:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("=" * 40)

print("\nDataset information:")
df.info()

print("=" * 40)
print("\nReview of the dataset:")
df.head()
df.sample(5, random_state=42)
```

```python
print("\nList of columns:")
df.columns.to_list()
```

Showing the number of missing values ​​in the dataset

```python
print("\nMissing values: ")
missing_values = df.isna().sum()
missing_values[missing_values>0]
```

Acquaintance with the column is of **great importance**, what values ​​are in it, extreme values ​​(minimum and maximum values) as well as the type of column.

```python
print("\nReview of column of main interes:")
print(df['priceUSD'].head(10))
print("*" * 40)
print("Minimal value:", df['priceUSD'].min())
print("Maximum value:", df['priceUSD'].max())
print("-" * 40)
print("Data type of column:", df['priceUSD'].dtype)
```

Here we make sure that the missing values ​​belong to one of the values ​​from the set `missing_like_values`.

```python
missing_like_values = ["NaN", "nan", "NULL", "null", "None", "none", "", " "]

for column in df.columns:
    if df[column].dtype == "object" or "float64" or "Int64":
        count = df[column].astype(str).str.strip().isin(missing_like_values).sum()
        if count > 0:
            print(f"{column}: {count}")
```

Overview of all `numeric columns`

```python
numeric_columns = [
    "mileage(kilometers)",
    "volume(cm3)",
    "year",
    "priceUSD"
]

df[numeric_columns].dtypes
df[numeric_columns].describe().round(2)
```

Overview of all `categorical_columns`

Here we checked all the unique values ​​found in the columns

```python
categorical_columns = [
    "make",
    "model",
    "condition",
    "fuel_type",
    "color",
    "transmission",
    "drive_unit",
    "segment"
]

for column in categorical_columns:
    print(column)
    print(df[column].nunique())
    print("-" * 40)
```

Here we got a detailed insight into all the values ​​found in the categorical columns (objects).

```python
for column in categorical_columns:
    print(df[column].value_counts(dropna=False))
    print("-" * 40)
```

Investigation of extreme values ​​by numerical columns.

```python
numeric_columns = df.select_dtypes(include="number").columns

print(numeric_columns)

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"\nColumn: {col}")
    print(f"Lower limit: {lower_bound:.2f}")
    print(f"Upper limit: {upper_bound:.2f}")
    print(f"Number of extreme values: {len(outliers)}")

    print(outliers[[col]].head(10))
```

Checking for duplicates

```python
duplicate_count = df.duplicated().sum()

print("Number of completely duplicated rows:", duplicate_count)
print(f"Percentage of completely duplicated rows: {duplicate_count / len(df) * 100:.2f}%")
```

Checking for ivalid values in numerical data.

```python
print("Negative prices:", (df["priceUSD"] < 0).sum())
print("Negative mileage:", (df["mileage(kilometers)"] < 0).sum())
print("Negative engine volume:", (df["volume(cm3)"] < 0).sum())
print("Invalid years:", (df["year"] <= 0).sum())

print("Minimum year:", df["year"].min())
print("Maximum year:", df["year"].max())

invalid_years = df[
    (df["year"] < 1900) |
    (df["year"] > 2026)
]

display(invalid_years)
```
